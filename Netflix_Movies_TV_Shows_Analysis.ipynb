```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Unknown")
df["duration"] = df["duration"].fillna("Unknown")
```

```python
import matplotlib.pyplot as plt
```

```python
df["type"].value_counts().plot(kind="bar")

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Number of Titles")
plt.show()
```

```python
country_counts = (
    df["country"]
    .str.split(", ")
    .explode()
    .value_counts()
)

country_counts.head(10)
```

```python
country_counts.head(10).plot(kind="bar")

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
genre_counts = (
    df["listed_in"]
    .str.split(", ")
    .explode()
    .value_counts()
)

genre_counts.head(10)
```

```python
genre_counts.head(10).plot(kind="bar")

plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
rating_counts = df["rating"].value_counts()

rating_counts
```

```python
rating_counts.head(10).plot(kind="bar")

plt.title("Top 10 Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
year_counts = df["release_year"].value_counts().sort_index()

year_counts.tail(20)
```

```python
year_counts.plot(kind="line", figsize=(12, 6))

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid()
plt.show()
```

```python
year_type_counts = df.groupby(["release_year", "type"]).size().unstack(fill_value=0)

year_type_counts.tail(20)
```

```python
year_type_counts.plot(kind="line", figsize=(12, 6))

plt.title("Movies vs TV Shows by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.legend(title="Type")
plt.grid()
plt.show()
```

```python
director_counts = df[df["director"] != "Unknown"]["director"].value_counts()

director_counts.head(10)
```

```python
director_counts.head(10).plot(kind="bar")

plt.title("Top 10 Directors on Netflix")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("netflix.db")

df.to_sql("netflix", conn, if_exists="replace", index=False)

print("Netflix data loaded into SQL successfully!")
```

```python
pd.read_sql_query("""
SELECT type, COUNT(*) AS total_titles
FROM netflix
GROUP BY type;
""", conn)
```

```python
pd.read_sql_query("""
SELECT COUNT(*) AS total_titles
FROM netflix;
""", conn)
```

```python
pd.read_sql_query("""
SELECT country, COUNT(*) AS total_titles
FROM netflix
GROUP BY country
ORDER BY total_titles DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql_query("""
SELECT rating, COUNT(*) AS total_titles
FROM netflix
GROUP BY rating
ORDER BY total_titles DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql_query("""
SELECT COUNT(*) AS total_titles
FROM netflix
WHERE release_year > 2018;
""", conn)
```

```python
pd.read_sql_query("""
SELECT director, COUNT(*) AS total_titles
FROM netflix
WHERE director != 'Unknown'
GROUP BY director
ORDER BY total_titles DESC
LIMIT 10;
""", conn)
```

```python
pd.read_sql_query("""
SELECT rating, type, COUNT(*) AS total_titles
FROM netflix
GROUP BY rating, type
ORDER BY total_titles DESC
LIMIT 15;
""", conn)
```
